# Python — String Methods
## Case Conversion · Stripping · replace() · split() & join()

**What this notebook covers**

1. **Case conversion** — `upper()`, `lower()`, `capitalize()`, `title()`
2. **Stripping characters** — `strip()`, `lstrip()`, `rstrip()` with custom character sets
3. **`replace()`** — deleting characters + the classic *swap trick* using a placeholder
4. **`split()` & `join()`** — breaking a string into a list and rebuilding it

> **Key idea for the whole notebook:** strings are **immutable** — every method returns a
> **new** string, the original never changes. To keep a change, reassign: `x = x.method()`.

---
## Section 1 — Case Conversion Methods

| Method | What it does | `"hello world !"` becomes |
|---|---|---|
| `upper()` | ALL letters to uppercase | `HELLO WORLD !` |
| `lower()` | all letters to lowercase | `hello world !` |
| `capitalize()` | only the **first letter of the string** uppercase, rest lowercase | `Hello world !` |
| `title()` | first letter of **every word** uppercase | `Hello World !` |

Non-letter characters (spaces, digits, `!`, `#`) are simply left untouched.

```python
x = " hello world !"

print(x.upper())    #  HELLO WORLD !
print(x.lower())    #  hello world !

print(x)            # original is UNCHANGED -> strings are immutable
```

### `capitalize()` vs `title()`

- `capitalize()` → only the very first character of the **whole string**.
- `title()` → first character of **each word**.

```python
x = "hello world !"

print(x.capitalize())   # Hello world !
print(x.title())        # Hello World !
```

## Section 2 — Stripping Characters

The `strip()` family removes unwanted characters from the **ends** of a string
(never from the middle).

- `lstrip(chars)` → cleans the **left** end
- `rstrip(chars)` → cleans the **right** end
- `strip(chars)` → cleans **both** ends
- With **no argument**, all three remove whitespace.

**How the `chars` argument works:** it is a **set of characters**, not a word.
Python keeps removing characters from that end *as long as* they belong to the set,
and stops at the first character that does not.

```python
x = " ###hello world !###*"

print(x.rstrip(" #*!"))   # ' ###hello world' -> only the RIGHT end is cleaned
```

Notice the left side still has `" ###"` — `rstrip()` never touches the left end.
Use `strip()` to clean both sides at once:

```python
print(x.strip(" #*!"))    # 'hello world' -> both ends cleaned
```

## Section 3 — `replace(old, new)`

Replaces **every** occurrence of `old` with `new` and returns a new string.
Replacing with an empty string `""` is the standard way to **delete** characters.

```python
x = "123-1234-157"

x = x.replace("-", "")   # delete every '-'
print(x)                 # 1231234157
```

### Class problem — swap `.` and `,`

**Task:** in the string below, every `.` must become `,` **and** every `,` must become `.`

**Why the direct way fails:** doing `replace(".", ",")` first turns *all* dots into commas —
now the old commas and the new commas are mixed together, and the second replace
`replace(",", ".")` converts **everything** into dots. The swap is destroyed.

**The trick:** park the dots in a **temporary placeholder** (`#` — any symbol *not present*
in the string), swap the commas, then convert the placeholder back.

```python
x = "welcome.to,vedam,for,learning.machine.learning,"

x = x.replace(".", "#")   # step 1: . -> #   (dots parked safely)
x = x.replace(",", ".")   # step 2: , -> .
x = x.replace("#", ",")   # step 3: # -> ,   (parked dots become commas)

print(x)                  # welcome,to.vedam.for.learning,machine,learning.
```

## Section 4 — `split()` and `join()`

These two are exact opposites:

- `s.split(separator)` → cuts the string at every separator → returns a **list** of pieces.
- `"glue".join(lst)` → joins the list items back into **one string**, placing the glue between them.

Watch the `join` syntax carefully — the **separator string comes first**, and the list goes
inside the brackets: `" -- ".join(lst)`.

```python
x = "welcome#to vedam for#   learning machine#learning"

lst = x.split("#")        # cut at every '#'
print(lst)
```

With a **custom separator**, `split("#")` cuts *only* at `#` — spaces inside the pieces are
kept as they are. (Compare with `split()` — no argument — which cuts at whitespace and also
eats the extra spaces.)

```python
new_str = " -- ".join(lst)   # rebuild with ' -- ' placed between the pieces
print(new_str)
```

## Quick Recap

| Method | Purpose |
|---|---|
| `upper()` / `lower()` | change case of all letters |
| `capitalize()` / `title()` | first letter of the string / of every word |
| `strip()` / `lstrip()` / `rstrip()` | remove characters from both ends / left / right |
| `replace(old, new)` | replace all occurrences; `replace(x, "")` deletes |
| `split(sep)` | string → list |
| `"sep".join(lst)` | list → string |

**Remember:** every string method returns a **new** string — reassign (`x = x.method()`)
to keep the result.
